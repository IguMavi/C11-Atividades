Questão 2)

```python
numero = int(input("Digite o número da tabuada: "))
inicio = int(input("Digite o início do intervalo: "))
fim = int(input("Digite o fim do intervalo: "))

if inicio > fim:
    inicio, fim = fim, inicio

print(f"\nTabuada do {numero}:")
for i in range(inicio, fim + 1):
    print(f"{numero} x {i} = {numero * i}")
```

```python
distancia = float(input("Digite a distância da viagem em Km: "))

if distancia <= 200:
    preco = distancia * 0.50
else:
    preco = distancia * 0.45

print(f"Preço da passagem: R$ {preco:.2f}")
```

Questão 5)

```python
numero = int(input("Digite um número entre 1000 e 9999: "))

while numero < 1000 or numero > 9999:
    print("Número inválido.")
    numero = int(input("Digite um número entre 1000 e 9999: "))

unidade = numero % 10
dezena = (numero // 10) % 10
centena = (numero // 100) % 10
milhar = numero // 1000

print("Unidade:", unidade)
print("Dezena:", dezena)
print("Centena:", centena)
print("Milhar:", milhar)
```

Questão 6)

```python
import math

numero = float(input("Digite um número decimal não negativo: "))

while numero < 0:
    print("Digite um número maior ou igual a zero.")
    numero = float(input("Digite um número decimal não negativo: "))

print("Raiz quadrada:", math.sqrt(numero))
print("Função teto:", math.ceil(numero))
print("Função chão:", math.floor(numero))
print("Parte inteira:", int(numero))
```

Questão 7)

```python
palavra = input("Digite uma palavra: ")

print("\nLetras:")
for letra in palavra:
    print(letra.upper())

vogais = "aeiou"
quantidade_vogais = sum(1 for letra in palavra.lower() if letra in vogais)
tem_a = "a" in palavra.lower()

print("\nQuantidade de vogais:", quantidade_vogais)

if tem_a:
    print('A letra "a" está presente na palavra.')
else:
    print('A letra "a" não está presente na palavra.')
```

```python
numero1 = float(input("Digite o primeiro número: "))
numero2 = float(input("Digite o segundo número: "))

adicao = numero1 + numero2
subtracao = numero1 - numero2
multiplicacao = numero1 * numero2
potencia = numero1 ** numero2

print("\nAdição:", adicao)
print("Subtração:", subtracao)
print("Multiplicação:", multiplicacao)

if numero2 != 0:
    print("Divisão:", numero1 / numero2)
    print("Resto da divisão:", numero1 % numero2)
else:
    print("Divisão: não é possível dividir por zero.")
    print("Resto da divisão: não é possível dividir por zero.")

print("Potência:", potencia)

if adicao % 2 == 0:
    print("O resultado da adição é par.")
else:
    print("O resultado da adição é ímpar.")
```
